### Primary Objective 
This project builds a machine learning classification model that **predicts National Park crowd levels as Low, Medium, High, or Extreme** by combining weather, wildfire, gas prices, and public interest trends.

### The system is designed to serve two primary purposes:
1. Assist travelers in trip planning
2. Support National Park management to allocate resources

### Machine Learning Justification
While seasonality provides a strong baseline pattern, deviations occur due to various factors. Machine learning models are well-suited to modeling these complex relationships.

### 4. Reviewing Fire Data
- This data was requested and downloaded from FIRMS: https://firms.modaps.eosdis.nasa.gov/download/ and loaded as a csv
- Data dictionary: https://www.earthdata.nasa.gov/data/tools/firms/active-fire-data-attributes-modis-viirs

````python
#keep only important rows
rowstokeep = ['latitude', 'longitude', 'acq_date', 'frp', 'type']
fire=fire0[rowstokeep].copy()

#remove type 1- active volcano
fire = fire[fire['type']!= 1]

#add date columns
fire['date'] = pd.to_datetime(fire['acq_date'])
fire['Year']= fire['date'].dt.year
fire['Month']= fire['date'].dt.month
fire['Day']= fire['date'].dt.day
fires= fire[['Year','Month', 'Day','latitude', 'longitude',  'frp']]


print(f' There are {round(fires.shape[0]/1000000,2)} million total fires in this dataset')
fires.head(3)
````

````python
#first get coordinates for each park

#_______________used Google Gemini for help to find coordinates of US parks from NPS API___________________________
import requests
#API_KEY = "KaElcDb5axwgiRzSlZFRlZhOy6wVNkUjj91MuEz6"
#url = f"https://developer.nps.gov/api/v1/parks?limit=500&api_key={API_KEY}"
#response = requests.get(url).json()
#park_list = []
#missing_park_codes = ["npsa", "redw"]
#for park in response["data"]:
#    # Filter for standard "National Park" designations
 #   code = park.get("parkCode", "").lower()
#    if "National Park" in park["designation"] or "National Parks" in park["designation"] or code in missing_park_codes:
#        park_list.append({
#                "park_code": park.get("parkCode", ""),
#                "park_name": park["fullName"],
#                "state": park["states"],
#                "Parklatitude": float(park["latitude"])
#                if park["latitude"]
#                else None,
#                "Parklongitude": float(park["longitude"])
#                if park["longitude"]
#                else None,})
#____________________________________________________________________________________________________________________

#save as csv
#df_nps_api = pd.DataFrame(park_list)
#df_nps_api.to_csv('nps_parks.csv', index=False)
parksinfo0 = pd.read_csv('data/nps_parks.csv')


#clean up list
#1- separate Sequoia and Kings Canyon
lat = parksinfo0.loc[parksinfo0['park_code']== 'seki','Parklatitude'].item()
lon = parksinfo0.loc[parksinfo0['park_code']== 'seki','Parklongitude'].item()
parksinfo = parksinfo0[parksinfo0['park_code'] != 'seki']
sequoia_row = {'park_code': 'sequ', 'park_name':'Sequoia National Park', 'state':'CA', 'Parklatitude':lat, 'Parklongitude':lon}
kings_row = {'park_code': 'kica', 'park_name':'Kings Canyon National Park', 'state':'CA', 'Parklatitude':lat, 'Parklongitude':lon}
new_rows = pd.DataFrame([sequoia_row, kings_row])
parksinfo = pd.concat([parksinfo, new_rows], ignore_index=True)

#2- change parks that are in 2 states to 1 primary state
parksinfo.loc[parksinfo['park_name'].str.contains('Yellowstone', case=False), 'state'] = 'WY'
parksinfo.loc[parksinfo['park_name'].str.contains('Death Valley', case=False), 'state'] = 'CA'
parksinfo.loc[parksinfo['park_name'].str.contains('Great Smoky', case=False), 'state'] = 'TN'

#3 - get rid of extra characters
parksinfo['park_name'] = parksinfo['park_name'].str.replace(' & Preserve', '', regex=False)
parksinfo['park_name'] = parksinfo['park_name'].str.replace('- ', '', regex=False)
parksinfo['park_name'] = parksinfo['park_name'].str.replace('Haleakalā', 'Haleakala', regex=False)
parksinfo['park_name'] = parksinfo['park_name'].str.replace('Hawaiʻi', 'Hawaii', regex=False)

#4 - capitalize codes
parksinfo['park_code'] = parksinfo['park_code'].str.upper()

#now have all 63 parks
print (f'There are {parksinfo.shape[0]} parks in this dataframe')
parksinfo.head(3)
````

````python
#_________________used Google Gemini for help with BallTree code__________________________________
# builds a 100km radius around each national park and identifies if each fire falls into a parks bound

#convert lat/loong coordinates to radians for each park & each fire
parks_coords = np.radians(parksinfo[['Parklatitude','Parklongitude']])
fires_coords = np.radians(fires[['latitude','longitude']])
#build BallTree
from sklearn.neighbors import BallTree
tree = BallTree(fires_coords, metric='haversine')
radius100 = 100/6371   # radians/ Earth is 6371 radians

#for each park, gives all fire records within radius
#output is 62 np array- each array gives the index of what fires are in that buffer
parktoFireIndex = tree.query_radius(parks_coords, r=radius100)
park_monthly_fire_counts = []
for parkidx, fireidx in enumerate(parktoFireIndex):
    nearby_fires = fires.iloc[fireidx]                                                     #assigns fires to each park to prepare to aggregate
    monthly = (nearby_fires.groupby(["Year","Month"]).agg(fire_count=("frp","size"),       #aggregates per month
             frp_sum=("frp","sum"),frp_mean=("frp","mean"),frp_max=("frp","max") ).reset_index())
    monthly["Park"] = parksinfo.iloc[parkidx]["park_name"]                                 #Adds parkname
    monthly["UnitCode"] = parksinfo.iloc[parkidx]["park_code"].upper()
    park_monthly_fire_counts.append(monthly)                                                #appends to master DF

fire_features = pd.concat(park_monthly_fire_counts,ignore_index=True)                       #concat each park together
fire_features = fire_features[['UnitCode','Park','Year', 'Month', 'fire_count', 'frp_sum', 'frp_mean', 'frp_max']]
#____________________________________________________________________________________________________________________
print(f' There are {fire_features.shape[0]} rows and {fire_features.shape[1]} columns')
fire_features.head(3)
````

#- created and saved a .csv file


from pytrends.request import TrendReq
import pandas as pd
import time

#-- Initialize PyTrends
pytrends = TrendReq(hl='en-US', tz=360)

#-- List of the 63 U.S. National Parks
parks = list(parksinfo['park_name'])
#parks = list(parksinfo['park_name'])[:2]

#-- Store results here
all_trends = []

for park in parks:
    print(park)

    try:
        pytrends.build_payload([park], timeframe='2010-01-01 2025-12-31', geo='US')
        data = pytrends.interest_over_time()
        if data.empty:
            print(f"No data for {park}")
            continue

        # Remove partial column
        if 'isPartial' in data.columns:
            data = data.drop(columns=['isPartial'])

        # Convert weekly data to monthly
        monthly = (data.resample('MS').mean().reset_index())
        monthly.columns = ['date', 'trend_score']
        monthly['park'] = park
        all_trends.append(monthly)

        # Avoid rate limits
        time.sleep(2)

    except Exception as e:
        print(f"Error for {park}: {e}")
        time.sleep(10)

#-- Combine all parks
trends_df = pd.concat(all_trends, ignore_index=True)
print(trends_df.head())

#-- Save
trends_df.to_csv("Pytrends.csv", index=False)

#### 7. Outlier and Anomaly Detection

````python
# Outlier Detection instead of removing outliers

#_______________Used Google Gemini_____________________________

def flag_park_anomalies(group, columns_to_check):
    """
    Calculates IQR fences per column for a single park group 
    and returns a boolean Series indicating if ANY column is an anomaly.
    """
    # Initialize a matrix of False values with the same index as the group
    is_anomaly_matrix = pd.DataFrame(False, index=group.index, columns=columns_to_check)
    
    for col in columns_to_check:
        q1 = group[col].quantile(0.25)
        q3 = group[col].quantile(0.75)
        iqr = q3 - q1
        
        lower_fence = q1 - 1.5 * iqr
        upper_fence = q3 + 1.5 * iqr
        
        # Flag rows where the column value is outside the fences
        is_anomaly_matrix[col] = (group[col] < lower_fence) | (group[col] > upper_fence)
    
    # If ANY column checked in this row is an anomaly, mark the row as True
    return is_anomaly_matrix.sum(axis=1)>=2

# Specify the columns that could indicate a historic anomaly
metrics_to_screen = ['RecreationVisits', 'frp_mean', 'temp_mean','snow', 'rain'] 

# Ensure data is sorted chronologically per park first
masterEDA = masterEDA.sort_values(['UnitCode', 'Year', 'Month']).reset_index(drop=True)

# Apply the IQR screening park-by-park
masterEDA['Is_Anomaly'] = masterEDA.groupby('UnitCode', group_keys=False).apply(
                            lambda x: flag_park_anomalies(x, metrics_to_screen))

# Convert the resulting boolean (True/False) into an integer flag (1/0) for the ML model
masterEDA['Is_Anomaly'] = masterEDA['Is_Anomaly'].astype(int)

print(f' The percent of anomolies is {round(masterEDA.Is_Anomaly.value_counts()[1]/len(masterEDA)*100,1)} %')
````

CONCLUSIONS:  Instead of removing these rows and losing historical context, a binary flag column (Is_Anomaly) was engineered so the machine learning models recognize these drops as isolated real-world events rather than normal seasonal trends

#### 1. Create Lags

````python
#---------------------Used Google Gemini to help with troubleshooting-----------------------------------

# Define feature groups
dynamic_features = ['RecreationVisits', 'fire_count', 'frp_sum', 'frp_mean', 'frp_max', 'trend_score', 'temp_mean', 'temp_max', 
                    'temp_min', 'rain', 'snow', 'cloud_cover', 'GasPrice']
weather_rolling_features = ['temp_mean', 'rain', 'snow']

# =====================================================================
# CREATE FULL GRID & IMPUTE MISSING MONTHS
# Ensure time columns are integers
master0['Year'] = master0['Year'].astype(int)
master0['Month'] = master0['Month'].astype(int)
# Generate a perfect, unbroken grid of every Park, Year, and Month combo
all_parks = master0['UnitCode'].unique()
all_years = range(master0['Year'].min(), master0['Year'].max() + 1)
all_months = range(1, 13)
perfect_grid = pd.MultiIndex.from_product([all_parks, all_years, all_months], names=['UnitCode', 'Year', 'Month']).to_frame().reset_index(drop=True)
masterP = pd.merge(perfect_grid, master0, on=['UnitCode', 'Year', 'Month'], how='left')
#Sort sequentially so time moves perfectly downward by 1-month steps
masterP = masterP.sort_values(['UnitCode', 'Year', 'Month']).reset_index(drop=True)
# 5. ffill based on month if there are missing months
for col in dynamic_features:
    masterP[col] = masterP.groupby(['UnitCode', 'Month'])[col].ffill()  # Group by park and month so July fills missing July
    masterP[col] = masterP.groupby(['UnitCode', 'Month'])[col].bfill()  # Back fill Trend Score
# =====================================================================
# Create Lags
# Re-sort by Park, then Month
masterP = masterP.sort_values(['UnitCode', 'Year', 'Month']).reset_index(drop=True)
for col in dynamic_features:
    masterP[f'{col}_lag_1'] = masterP.groupby('UnitCode')[col].shift(1)   # A. 1-Month Lag (Immediate momentum)
    masterP[f'{col}_lag_12'] = masterP.groupby('UnitCode')[col].shift(12) # B. 12-Month Lag (Same month, previous year baseline)
for col in weather_rolling_features:
    # Built off lag_1 to prevent data leakage
    masterP[f'{col}_roll_mean_3'] = (masterP.groupby('UnitCode')[f'{col}_lag_1'] # C. 3-Month Chronological Rolling Window (Cumulative environmental trends)
                                         .rolling(window=3, min_periods=1).mean().reset_index(level=0, drop=True))
# =====================================================================
# FINAL CLEANUP & SORTING 
# Re-sort by Park, Year, then Month
masterP = masterP.sort_values(['UnitCode', 'Year', 'Month']).reset_index(drop=True)
# Drop the first 12 incomplete baseline rows of EACH park
dropped_groups = [group.iloc[12:] for _, group in masterP.groupby('UnitCode')]
masterP = pd.concat(dropped_groups, ignore_index=True)
print(f"Feature engineering complete! Shape of master dataframe: {masterP.shape} -- which = 63x12x20 - 12*63 = {63*12*20- (12*63)}")
````

#### 2. Assign Crowd Tiers relative to each park

````python
#---------------------Used Google Gemini to help with troubleshooting-----------------------------------

def assign_crowd_tiers(fulldata, split_year=2022, visit_col='RecreationVisits',
                               lag_col='RecreationVisits_lag_12', year_col='Year', saved_bins=None):
    df = fulldata.copy()
    quantiles = [0.0, 0.25, 0.65, 0.9, 1.0]
    labels = [0, 1, 2, 3]
    trained_bins = {}     
    
    # 1. Isolate historical records up to the split year to calculate thresholds
    #split year is included in training then
    train_history = df[df[year_col] <= split_year]
    
    # 2. Build the bin boundaries for every park using aggregation
    if saved_bins is None:
        for park in df['UnitCode'].unique():
            park_visits = train_history[train_history['UnitCode'] == park][visit_col]
            
            if len(park_visits) == 0:
                continue
                
            raw_bounds = np.quantile(park_visits.to_numpy(), quantiles)
            raw_bounds[0] = -1
            raw_bounds[-1] = np.inf
            trained_bins[park] = raw_bounds.tolist()
    else:
        trained_bins = saved_bins

    # 3. Apply the bins cleanly using a single helper function to prevent Index conflicts
    def compute_tier(row, column_to_cut):
        park = row['UnitCode']
        if park not in trained_bins:
            return np.nan
        val = row[column_to_cut]
        if pd.isna(val):
            return np.nan
        
        # Use pd.cut directly on the isolated scalar array
        b = trained_bins[park]
        return int(pd.cut([val], bins=b, labels=labels, include_lowest=True)[0])

    # 4. Safely apply the function row-by-row across the unified dataset
    df['CrowdTier'] = df.apply(lambda r: compute_tier(r, visit_col), axis=1)
    
    # 5. Fix the lag missing gaps chronologically per park before cutting baseline predictions
    df = df.sort_values(by=['UnitCode', 'Year', 'Month']).reset_index(drop=True)
    df[lag_col] = df.groupby('UnitCode')[lag_col].ffill().fillna(0)
    
    df['Baseline_Prediction'] = df.apply(lambda r: compute_tier(r, lag_col), axis=1)
    
    # 6. Drop any unmapped rows (e.g., parks with no history before 2022) and cast types
    df = df.dropna(subset=['CrowdTier', 'Baseline_Prediction']).copy()
    df['CrowdTier'] = df['CrowdTier'].astype(int)
    df['Baseline_Prediction'] = df['Baseline_Prediction'].astype(int)
    
    return df, trained_bins
````

#### 4. Custom Transformers

````python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler, FunctionTransformer
from sklearn.impute import SimpleImputer
from sklearn.base import BaseEstimator, TransformerMixin
````

````python
#seasonal imputation past 3 years for that park for that month
class ParkSeasonalImputer(BaseEstimator, TransformerMixin):
    def __init__(self, columns):
        self.columns = columns
        self.park_col = 'UnitCode'
        self.month_col = 'Month'

    def fit(self, X, y=None):
        X = X.copy()
        self.global_means = X[self.columns].mean()            #for fallback impuatation
        self.lookup = {}                                      #create dictionary
        for col in self.columns:                              #compute park/month avg 
            self.lookup[col] = (X.groupby([self.park_col, self.month_col])[col]
                                .mean())
        return self

    def transform(self, X):
        for col in self.columns:
            X = X.merge(self.lookup[col].rename(col + "_imp"),
                    left_on=[self.park_col, self.month_col], right_index=True, how="left")
            X[col] = X[col].fillna(X[col + "_imp"])
            X = X.drop(columns=[col + "_imp"])
                # fallback global mean
            X[col] = X[col].fillna(self.global_means[col])
        return X

    def get_feature_names_out(self, input_features= None):
        return input_features
````

````python
class UnitCodeTargetEncoder(BaseEstimator, TransformerMixin):

    def __init__(self,
                 unit_col='UnitCode',
                 visit_col='RecreationVisits',
                 output_col='UnitCodeTE'):
        self.unit_col = unit_col
        self.visit_col = visit_col
        self.output_col = output_col


    def fit(self, X, y=None):
        self.feature_names_in_ = X.columns
        self.global_median_ = X[self.visit_col].median()
        self.mapping_ = (X.groupby(self.unit_col)[self.visit_col].median())
        return self

    def transform(self, X):
        X = X.copy()
        X[self.output_col] = (X[self.unit_col].map(self.mapping_).fillna(self.global_median_) )
        return X

    def get_feature_names_out(self, input_features= None):
        if input_features is None:
            input_features=self.feature_names_in_
        features = list(input_features)
        if self.output_col not in features:
            features.append(self.output_col)
        return np.array(features)
````

````python
class CyclicalMonthEncoder(BaseEstimator, TransformerMixin):
    def fit(self, X, y=None):
        return self

    def transform(self, X):
        X_out = pd.DataFrame(X).copy()
        X_out['Month_sin'] = np.sin(2 * np.pi * (X_out['Month'] - 1) / 12)
        X_out['Month_cos'] = np.cos(2 * np.pi * (X_out['Month'] - 1) / 12)
        return X_out.drop(columns='Month')

    def get_feature_names_out(self, input_features=None):
        return np.array(['Month_sin','Month_cos'])
    
````

````python
class DropColumnsTransformer(BaseEstimator, TransformerMixin):

    def __init__(self, columns):
        self.columns = columns

    def fit(self, X, y=None):
        # remember original columns
        self.feature_names_in_ = list(X.columns)
        return self

    def transform(self, X):
        X = X.copy()
        return X.drop(columns=self.columns, errors='ignore')

    def get_feature_names_out(self, input_features=None):
        if input_features is None:
            input_features = self.feature_names_in_
        output_features = [col for col in input_features if col not in self.columns ]
        return np.array(output_features)
````

````python
#Split into X & Y
monthfeatures = ['Month']
firefeatures = ['fire_count_lag_1'] 
numfeatures = ['Baseline_Prediction','relative_visits_lag_12','relative_visits_lag_1', 'trend_score_lag_1',
               'is_pandemic', 'temp_mean_roll_mean_3', 'rain_roll_mean_3', 'snow_roll_mean_3','trend_velocity','comfort_index'] 
TBDfeatures = [ 'RecreationVisits', 'UnitCode']
trainfeatures =  monthfeatures + firefeatures + numfeatures + TBDfeatures

X_train = train_df[trainfeatures]
y_train = train_df['CrowdTier'].copy()

X_test = test_df[trainfeatures] 
y_test = test_df['CrowdTier'].copy()

print(f'The percent held for testing is {round(X_test.shape[0]/masterP.shape[0]* 100 , 1)}% of the data. The training set is {X_train.shape[0]} rows and the test set is {X_test.shape[0]} rows. There are {X_test.shape[1]} columns '  )
````

#### 6. Make Pipeline

````python
fire_pipeline = Pipeline([('log', FunctionTransformer(np.log1p, feature_names_out="one-to-one")),
                          ('scale', StandardScaler() ) ])

preprocessor = Pipeline([ ("seasonal_imputer", ParkSeasonalImputer(columns=numfeatures)),
                          ("unit_encoder", UnitCodeTargetEncoder()),
                          ("drop_cols", DropColumnsTransformer(columns= TBDfeatures)),
                          ("preprocess", ColumnTransformer([("fire", fire_pipeline, firefeatures),
                                                            ('month', CyclicalMonthEncoder(), monthfeatures ),
                                                            ('num', make_pipeline(StandardScaler()), numfeatures + ["UnitCodeTE"]) ]))])
````

#### Base Line Model
The base line model simply predicts the crowd level by using the **prior year's crowd level** for that specific park & month. 
So if we want to predict July 2025 crowd level, it simply outputs July 2024 crowd level tier. This baseline prediction was created in the assign_crowd_tier function. 

````python
from sklearn.metrics import classification_report, f1_score

baselinef1train_MAC = f1_score(y_train, train_df['Baseline_Prediction'], average = 'macro')
baselinef2train_WEI = f1_score(y_train, train_df['Baseline_Prediction'], average = 'weighted')

print(f"BASELINE PERFORMANCE (2006-2022 Train Set)")
print(f"Overall Weighted F1 Score: {baselinef2train_WEI:.4f}")
print(f"Overall Macro F1 Score:    {baselinef1train_MAC:.4f}")

print("=" * 60)
print('----------Classification Report of Training Set----------')
print(classification_report(y_train, train_df['Baseline_Prediction'], target_names=['Low', 'Medium', 'High', 'Extreme'] ))
````

After seeing the high accuracy of the Baseline Model, I decided to add the Baseline Tier as a feature into the machine learning models. This changes the objective of the project as an optimization and to use the baseline as a foundation and then make adjustments to address edge-cases. 

#### Machine Learning Models

````python
from sklearn.model_selection import RandomizedSearchCV, TimeSeriesSplit
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import classification_report, f1_score
from sklearn.utils.class_weight import compute_class_weight
import joblib
````

````python
#macro and weighted F1

y_train = train_df['CrowdTier'].astype(int).values

#-------------------Define Validation parameters------------------
classes = np.unique(y_train)
weights = compute_class_weight(class_weight='balanced', classes=classes, y=y_train)
class_weights_dict = dict(zip(classes, weights))

models = { 'LogisticRegression': LogisticRegression(class_weight=class_weights_dict, max_iter=1000, random_state=42),
           'RandomForest': RandomForestClassifier(class_weight=class_weights_dict, random_state=42),
           'XGBoost': XGBClassifier(objective='multi:softprob', num_class=4, random_state=42)}

param_grids = { 'LogisticRegression': {'classifier__C': [1.0, 10.0]},
                'RandomForest': {'classifier__n_estimators': [250, 350],
                                 'classifier__max_depth': [14, 18],     
                                 'classifier__min_samples_leaf': [3, 6],     
                                 'classifier__max_features': ['sqrt']},
                    'XGBoost': {'classifier__n_estimators': [200, 300],
                                'classifier__max_depth': [5, 7],               
                                'classifier__learning_rate': [0.04, 0.06], 
                                'classifier__colsample_bytree': [0.7, 0.85], 
                                'classifier__subsample': [0.8, 0.9]}}

#-------------------Time Series Validation------------------
print("Starting Tuning for all 3 models... \n"); print("-" * 70)

tscv = TimeSeriesSplit(n_splits=3)
best_pipelines = {}

for model_name, model_obj in models.items():
    full_pipeline = Pipeline([ ('preprocessor', preprocessor), 
                               ('classifier', model_obj)])
    
    search = RandomizedSearchCV(estimator=full_pipeline,
                                param_distributions=param_grids[model_name],
                                n_iter=10 if model_name != 'LogisticRegression' else 4, 
                                scoring={ 'macro_f1': 'f1_macro','weighted_f1': 'f1_weighted'},
                                refit='macro_f1', 
                                cv=tscv,random_state=42,n_jobs=-1)
    search.fit(X_train, y_train)
    best_pipelines[model_name] = search.best_estimator_
    best_idx = search.best_index_
    cv_macro = search.cv_results_['mean_test_macro_f1'][best_idx]
    cv_weighted = search.cv_results_['mean_test_weighted_f1'][best_idx]
    
    print(f"Optimal {model_name} Cross-Validation:")
    print(f"   -> Mean CV Weighted F1: {cv_weighted:.4f}")
    print(f"   -> Mean CV Macro F1:    {cv_macro:.4f}")
    print(f"   -> Metric Delta (Gap):  {abs(cv_weighted - cv_macro):.4f}"); print("-" * 70)

#Compare to Baseline Results
baselinef1train_MAC = f1_score(y_train, train_df['Baseline_Prediction'], average = 'macro')
baselinef2train_WEI = f1_score(y_train, train_df['Baseline_Prediction'], average = 'weighted')
print("Baseline Training Set Results")
print(f"Overall Weighted F1 Score: {baselinef2train_WEI:.4f}")
print(f"Overall Macro F1 Score:    {baselinef1train_MAC:.4f}")
````

**Conclusion**: XGBoost is the optimal model based on F1 scores. It exceeds the baseline model and the difference between
weighted and macro F1 tells us the model is not biased across classes.

````python
print('XG Boost model')
preds = best_pipelines['XGBoost'].predict(X_train)
print(classification_report(y_train, preds, target_names=['Low', 'Medium', 'High', 'Extreme'] ))

print('random forest')
predsRF = best_pipelines['RandomForest'].predict(X_train)
print(classification_report(y_train, predsRF, target_names=['Low', 'Medium', 'High', 'Extreme'] ))
````

**Conclusion**:The results of the output show an impressive 95% recall for the Random Forest model which signals signs
of overfitting. The XG Boost is the more trustworthy model for new data.

````python
from sklearn.calibration import calibration_curve
from sklearn.metrics import brier_score_loss

models_to_compare = ['RandomForest', 'XGBoost']
colors = ['#2ca02c', '#ff7f0e', '#d62728', '#1f77b4','#9467bd']
fig, axes = plt.subplots(1, 2, figsize=(16, 7), sharex=False)
sns.set_theme(style="whitegrid")

# Create  Probability Calibration Graphs
for idx, model_name in enumerate(models_to_compare):
    ax = axes[idx]
    ax.plot([0, 1], [0, 1], linestyle="--", color="gray", label="Perfect Calibration") # Perfect calibration for reference
    holdout_probs = best_pipelines[model_name].predict_proba(X_train)                  # Get probabilities for each model
    detected_classes = best_pipelines[model_name].named_steps['classifier'].classes_    
    # Inner loop computing curve metrics for each crowd tier class
    for i, class_name in enumerate(detected_classes):
        class_probs = holdout_probs[:, i]
        y_train_binary = (y_train == class_name).astype(int)     # Build binary evaluation target mapping                                                      
        prob_true, prob_pred = calibration_curve(y_train_binary, class_probs, n_bins=5, strategy='uniform')
        brier_score = brier_score_loss(y_train_binary, class_probs)
        ax.plot(prob_pred, prob_true, marker="o", linewidth=2,   # Plot class curve on the active axis
                color=colors[i % len(colors)], label=f"Class {class_name} (Brier: {brier_score:.4f})")
    ax.set_title(f"{model_name} Reliability Diagram", fontweight="bold", fontsize=13)
    ax.set_xlabel("Mean Predicted Probability", fontsize=11)
    if idx == 0:
        ax.set_ylabel("Fraction of True Outcomes (Actual Frequency)", fontsize=11) 
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.legend(loc="lower right", fontsize=9)
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

**Conclusion**:This plot shows that the XGBoost model is more aligned with a
perfect calibration demonstrating more trustworthy probability estimates.
The S shape of the Random Forest means this model tends to be over
confident and under confident making it less reliable

A lookup table will be created using 2025 data as a historical baseline to be used as predictions for 2026 crowd levels. 

````python
finaldf= labeled_df.copy()
df2025 = finaldf[finaldf['Year']==2025].copy()          #filtered df of just 2025

lookuprows= []
parks63= finaldf['UnitCode'].unique()                    #list of each unique park
parks63 

for park in parks63:
    p_2025= df2025[df2025['UnitCode']==park]             #filters for just that park in 2025        
    for month in range (1,13):
        lag_1_month = 12 if month ==1 else month - 1      #look at Dec if target month is Jan.
        row2025target = p_2025[p_2025['Month']== month].iloc[0]
        row2025_lag1 =  p_2025[p_2025['Month']== lag_1_month].iloc[0]
        rowdict = { #columns needed for model
                    "UnitCode": park,
                    "Month": month,
                    "is_pandemic": 0,
                    #1 month lags (using 12 month lag for Jan)
                    "fire_count_lag_1": row2025target["fire_count_lag_1"] if month!= 1 else row2025_lag1['fire_count_lag_12'],
                    "relative_visits_lag_1": row2025target["relative_visits_lag_1"] if month!= 1 else row2025_lag1['relative_visits_lag_12'],
                    "trend_score_lag_1": row2025target["trend_score_lag_1"]if month!= 1 else row2025_lag1['trend_score_lag_12'] ,
                    #12 month lag
                    "relative_visits_lag_12": row2025target["relative_visits_lag_12"],
                   "trend_score_lag_12": row2025_lag1["trend_score_lag_12"],
                    #baseline and rolling weather
                    "Baseline_Prediction": row2025target["Baseline_Prediction"],
                    "temp_mean_roll_mean_3": row2025target["temp_mean_roll_mean_3"],
                    "rain_roll_mean_3": row2025target["rain_roll_mean_3"],
                    "snow_roll_mean_3": row2025target["snow_roll_mean_3"],
                    #extra columns needed for insights
                    "fire_count": row2025target["fire_count"] ,
                    "temp_max": row2025target["temp_max"] ,
                    "temp_min": row2025target["temp_min"],
                    "trend_score": row2025target["trend_score"] ,
                    "firecountavg": finaldf[(finaldf['UnitCode']== park) & (finaldf['Month']== month)]['fire_count'].mean()}
        lookuprows.append(rowdict)

lookupdf = pd.DataFrame(lookuprows)
lookupdf['trend_velocity'] = lookupdf['trend_score_lag_1'] - lookupdf['trend_score_lag_12']
lookupdf['comfort_index'] = ((lookupdf['temp_mean_roll_mean_3'] - 70).abs() * 0.1 + (lookupdf['rain_roll_mean_3'] * 2.0)) 
lookupdf = pd.merge(left= lookupdf, right= parksinfo[['park_code','park_name','Parklatitude','Parklongitude']],
                  left_on ='UnitCode', right_on= 'park_code', how = 'left' ).drop(columns= {'park_code'})
lookupdf = lookupdf.rename(columns = {'Parklatitude':'latitude','Parklongitude':'longitude'})
lookupdf.to_csv("app/park_lookup_data.csv", index=False)
lookupdf.head(3)
````
